# **Dataset: La Liga Match Data from 2014 - 2020**


https://www.kaggle.com/datasets/sanjeetsinghnaik/la-liga-match-data

La Liga is by far one of the world’s most entertaining leagues. They have some of the best managers, players and fans! But, what makes it truly entertaining is the sheer unpredictability. There are 6 equally amazing teams with a different team lifting the trophy every season. Not only that, the league has also witnessed victories from teams outside of the top 6. So, let us analyze some of these instances.

So far, the implementation of statistics into soccer has been positive. Teams can easily gather information about their opponents and their tactics. This convenience allows managers to create well-thought out game plans that suit their team, maximize opponents' weaknesses, and increase their chances of winning.

A goal is scored when the whole of the ball passes over the goal line, between the goalposts and under the crossbar, provided that no offence has been committed by the team scoring the goal. If the goalkeeper throws the ball directly into the opponents' goal, a goal kick is awarded.

THE TIME OF SEASON/MOTIVATION: While a club battling for a league title is going to be hungry for a win, as is a side that is fighting to stay up, a club that has already won the title or has already been relegated is unlikely to work as hard, and often rest players as well.
THE REFEREE: Of course, when referee's send players off it make a massive impact on a match, but even if he is just awarding a yellow card then it can affect the outcome of the game as the player booked is less likely to go in as hard for the rest of the match.

SUBSTITUTES: The whole point of substitutes is for them to be able to come on and impact a match. Subs not only bring on a fresh pair of legs that are less tired than starters and more likely to track back and push forward, but can also play crucial roles in the formation of a team.

MIND GAMES/MANAGERS: Playing mind games has almost become a regular routine for top level managers, and rightly so. Just a simple mind game can do so much to impact a match, a good example coming from Sir Alex Ferguson.

FORM/INJURIES/FIXTURES: A team on better form is more likely to win a match than if they have been on a poor run of form, while a team in the middle of a condensed run of fixtures is less likely to win than a well rested team.

# **Import Library**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

# **Memuat Dataset**

```python
# Load Dataset
df = pd.read_csv('combined_data_laliga.csv')

print("\nDataset Berhasil Dimuat")
print("Shape: ", df.shape)
print(df.head())
```

# **Exploratory Data Analysis (EDA)**


```python
# Buat kolom FTR (Full Time Result)
def get_ftr(row):
    if row['Home Team Goals Scored'] > row['Away Team Goals Scored']:
        return 'H'  # Menang Kandang
    elif row['Home Team Goals Scored'] < row['Away Team Goals Scored']:
        return 'A'  # Menang Tandang
    else:
        return 'D'  # Seri

df['FTR'] = df.apply(get_ftr, axis=1)

# Melihat ringkasan statistik dari kolom numerik
print("Ringkasan Statistik Numerik:")
display(df.describe())

# Mengecek jumlah nilai yang kosong (Missing Values) di setiap kolom
print("Jumlah Missing Values per Kolom:")
display(df.isnull().sum())

# Visualisasi Target Prediksi
kolom_target = 'FTR'

if kolom_target in df.columns:
    plt.figure(figsize=(8, 5))
    sns.countplot(x=kolom_target, data=df, palette='viridis')
    plt.title('Distribusi Hasil Pertandingan (Menang Kandang, Seri, Menang Tandang)')
    plt.xlabel('Hasil Pertandingan')
    plt.ylabel('Jumlah Pertandingan')
    plt.show()
else:
    print(f"Kolom '{kolom_target}' tidak ditemukan. Silakan cek hasil df.info() sebelumnya untuk menyesuaikan nama kolom hasil pertandingan.")

# Melihat korelasi antar fitur statistik numerik
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap Korelasi Fitur Pertandingan')
plt.show()
```

# **Data Preprocessing**

```python
# Eliminasi Data Duplikat
observasi_awal = df.shape[0]
df.drop_duplicates(inplace=True)
print(f"Jumlah observasi duplikat yang tereliminasi: {observasi_awal - df.shape[0]} baris.")

# Mencegah Data Bocor (Leakage)
kolom_penyebab_bocor = ['Home Team Goals Scored', 'Away Team Goals Scored']
kolom_untuk_dihapus = [col for col in kolom_penyebab_bocor if col in df.columns]
df.drop(columns=kolom_untuk_dihapus, inplace=True, errors='ignore')
print(f"Atribut penyebab kebocoran data (Data Leakage) telah dihapus: {kolom_untuk_dihapus}")

# Penanganan Nilai Kosong (Missing Values)
kolom_numerik = df.select_dtypes(include=[np.number]).columns
imputer = SimpleImputer(strategy='median')
df[kolom_numerik] = imputer.fit_transform(df[kolom_numerik])
print("Nilai kosong pada atribut numerik telah ditangani menggunakan metode Median Imputation.")

# Representasi Data Kategorikal (Encoding)
le_team = LabelEncoder()
le_target = LabelEncoder()

# Proses encoding pada nama entitas tim (asumsi atribut bernama 'Home Team' dan 'Away Team')
if 'Home Team' in df.columns and 'Away Team' in df.columns:
    seluruh_entitas_tim = pd.concat([df['Home Team'], df['Away Team']]).unique()
    le_team.fit(seluruh_entitas_tim)
    df['Home Team'] = le_team.transform(df['Home Team'])
    df['Away Team'] = le_team.transform(df['Away Team'])

# Proses encoding pada variabel target klasifikasi FTR (Full Time Result)
if 'FTR' in df.columns:
    df['FTR'] = le_target.fit_transform(df['FTR'])
    print("Pemetaan Representasi Target (FTR):", dict(zip(le_target.classes_, le_target.transform(le_target.classes_))))

# Pemisahan Atribut Fitur (X) dan Variabel Target (y)
X = df.drop(columns=['FTR'])
y = df['FTR']

# Identifikasi dan hapus kolom non-numerik yang tidak dapat di-scale
kolom_non_numerik_untuk_dihapus = ['Score', 'Half Time Score']
X = X.drop(columns=[col for col in kolom_non_numerik_untuk_dihapus if col in X.columns], errors='ignore')

# Standardisasi Atribut (Feature Scaling)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Pencegahan Kebocoran Data (Data Leakage) dan Pembersihan Atribut Non-Prediktif
kolom_penyebab_bocor = [
    'Home Team Goals Scored', 'Away Team Goals Scored',
    'Home Team Goals Conceeded', 'Away Team Goals Conceeded',
    'Unnamed: 0' # Sekalian menghapus kolom indeks
]
df.drop(columns=[col for col in kolom_penyebab_bocor if col in df.columns], inplace=True, errors='ignore')

# Mengonversi matriks hasil standardisasi kembali menjadi struktur DataFrame
X_final = pd.DataFrame(X_scaled, columns=X.columns)

print("\nProses Data Preprocessing telah selesai dieksekusi.")
print(f"Dimensi matriks data yang siap untuk fase pelatihan: {X_final.shape}")
print("\nCuplikan Data Preprocessed:")
display(X_final.head())
```
